**Check the Users Schema Privileges**

```sql
SELECT
    grantee,
    privilege_type,
    catalog_name,
    schema_name
FROM system.information_schema.schema_privileges
WHERE catalog_name = 'workspace'
  AND schema_name = '03_gold'
  AND grantee IN ('data_analysts', 'data_engineers');
```

**Data Analyst - Read-Only Access to Silver and Gold Tables**

```sql
GRANT USE CATALOG ON CATALOG workspace TO `data_analysts`;

GRANT USE SCHEMA ON SCHEMA workspace.02_silver TO `data_analysts`;

GRANT SELECT ON SCHEMA workspace.02_silver TO `data_analysts`;

GRANT USE SCHEMA ON SCHEMA workspace.03_gold TO `data_analysts`;

GRANT SELECT ON SCHEMA workspace.03_gold TO `data_analysts`;
```

**Data Engineer - Read, Write, and Update Access to All Tables**

```sql
GRANT USE CATALOG ON CATALOG workspace TO `data_engineers`;
GRANT USE SCHEMA, CREATE SCHEMA ON CATALOG workspace TO `data_engineers`;

GRANT CREATE TABLE, CREATE MATERIALIZED VIEW, CREATE VOLUME,
      SELECT, MODIFY, READ VOLUME, WRITE VOLUME
  ON SCHEMA workspace.01_bronze TO `data_engineers`;

GRANT CREATE TABLE, CREATE MATERIALIZED VIEW, CREATE VOLUME,
      SELECT, MODIFY, READ VOLUME, WRITE VOLUME
  ON SCHEMA workspace.02_silver TO `data_engineers`;
  
GRANT CREATE TABLE, CREATE MATERIALIZED VIEW, CREATE VOLUME,
      SELECT, MODIFY, READ VOLUME, WRITE VOLUME
  ON SCHEMA workspace.03_gold TO `data_engineers`;
```
